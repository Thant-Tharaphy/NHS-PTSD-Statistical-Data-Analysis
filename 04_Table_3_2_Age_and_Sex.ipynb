```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = "APMS23-24-Posttraumatic-Stress-Disorder-data-tables.xlsx"
```

```python
# Load Table 3.2 without assuming where the headers are

table_3_2 = pd.read_excel(
    file_path,
    sheet_name="Table 3.2",
    header=None
)

table_3_2
```

```python
# Extract the percentage observation section
age_observed = table_3_2.iloc[9:20].copy()

age_observed
```

```python
# Create a list that will store the gender for each row
gender_labels = []

# Start without a gender
current_gender = None

# Read each value in the first column
for value in age_observed[0]:

    # If we find a gender label, remember it
    if value in ["Males", "Females", "All adults3"]:
        current_gender = value

    # Give the current gender to this row
    gender_labels.append(current_gender)

# Add reconstructed gender column
age_observed["gender"] = gender_labels

age_observed
```

```python
# Keep only the actual observation rows
age_observed = age_observed[
    age_observed[0].isin([
        "Trauma experienced",
        "PTSD screen positive"
    ])
].copy()

age_observed
```

```python
# Standardise gender labels
age_observed["gender"] = age_observed["gender"].replace({
    "Males": "Men",
    "Females": "Women",
    "All adults3": "All adults"
})

age_observed
```

```python
# Create age group labels
age_groups = [
    "16-24", "25-34", "35-44", "45-54",
    "55-64", "65-74", "75+", "Total"
]

# Create new column names
new_columns = []

for age in age_groups:
    new_columns.append(f"{age}_2014")
    new_columns.append(f"{age}_2023/4")

new_columns
```

#### We're using a loop because every age group follows exactly the same pattern 


age → 2014 → 2023/4 → next age → 2014 → 2023/4

```python
# Rename the 16 percentage columns
age_observed.columns = (
    ["measure"]
    + new_columns
    + ["gender"]
)

age_observed
```

```python
# Reshape from wide format to long format
age_long = age_observed.melt(
    id_vars=["gender", "measure"],
    var_name="age_year",
    value_name="percentage"
)

age_long
```

#### Next we'll split age_year into two proper variables
age_group + survey_year.

```python
# Split age_year into age_group and survey_year
age_long[["age_group", "survey_year"]] = (
    age_long["age_year"].str.split("_", expand=True)
)

age_long
```

Algorithm note 

We use `.str.split()` to separate the combined `age_year` column into two individual columns: `age_group` and `survey_year`.

- `age_long["age_year"]` selects the column containing combined age and year labels.
- `.str.split("_")` splits each value at the underscore `_`.
- `expand=True` returns the separated values as individual columns.
- `age_long[["age_group", "survey_year"]]` assigns the results to two new columns.

Observations:

- `16-24_2014` becomes `16-24` and `2014`.
- `25-34_2023/4` becomes `25-34` and `2023/4`.
- Each row now contains a separate age group and survey year.
- The original percentages remain unchanged.
- The number of rows remains the same.


Why?

Separating age groups and survey years allows us to filter, group, and compare percentages independently by age and time.


age_group   = 16-24

survey_year = 2023/4

```python
# let's tidy the final percentage dataset

# Remove temporary age_year column
age_long = age_long.drop(columns=["age_year"])

# Arrange columns in a clear order
age_long = age_long[
    [
        "gender",
        "measure",
        "age_group",
        "survey_year",
        "percentage"
    ]
]

age_long
```


Algorithm note 

We use `.drop()` and column selection to remove the temporary column and organise the dataset into a clean structure.

- `age_long.drop(columns=["age_year"])` removes the temporary age_year column.
- The age_year column is no longer needed because we already separated it into age_group and survey_year.
- `age_long[[...]]` selects and rearranges the columns in the specified order.
- `age_long =` saves the cleaned DataFrame.

Observations:

- The temporary age_year column is removed.
- Five columns remain: gender, measure, age_group, survey_year, and percentage.
- Each row represents one gender, measure, age group, and survey year.
- The percentage values remain unchanged.
- The expected dataset contains 96 rows and 5 columns.

Why?

Removing unnecessary columns and organising variables makes the dataset easier to understand, validate, filter, and analyse.

#### PDST Screening Positive Visualization by age groups

```python
# Remove Total so the chart shows age groups only
age_ptsd_chart = age_ptsd[
    age_ptsd["age_group"] != "Total"
].copy()

# Set age groups in the correct order
age_order = [
    "16-24", "25-34", "35-44",
    "45-54", "55-64", "65-74", "75+"
]

age_ptsd_chart["age_group"] = pd.Categorical(
    age_ptsd_chart["age_group"],
    categories=age_order,
    ordered=True
)
```


Algorithm note - Preparing age groups for PTSD screening visualisation

We use conditional filtering and `pd.Categorical()` to prepare the PTSD screening dataset for visualisation.

- `age_ptsd["age_group"] != "Total"` selects all rows except the Total category.
- `.copy()` creates an independent DataFrame.
- `age_ptsd_chart` stores the filtered dataset.
- `age_order` defines the seven age groups from youngest to oldest.
- `pd.Categorical()` converts the age_group column into an ordered categorical variable.
- `categories=age_order` specifies the correct age group sequence.
- `ordered=True` tells Pandas that the categories have a meaningful order.

Observations:

- The Total category is removed.
- Seven age groups remain, from 16-24 to 75+.
- Both survey years and all available gender categories remain.
- The percentage values remain unchanged.
- Age groups are treated as ordered categories rather than ordinary text.



```python
# Clean Seaborn style
sns.set_theme(
    style="whitegrid",
    font="Arial"
)

# Create separate chart for each gender
g = sns.relplot(
    data=age_ptsd_chart,
    x="age_group",
    y="percentage",
    hue="survey_year",
    col="gender",
    kind="line",
    marker="o",
    linewidth=2.5,
    markersize=8,
    height=5,
    aspect=1
)

# Labels and titles
g.set_axis_labels(
    "Age Group",
    "PTSD Screen Positive (%)"
)

g.set_titles("{col_name}")

g.fig.suptitle(
    "PTSD Screening by Age Group: 2014 vs 2023/4",
    fontsize=15,
    fontweight="bold",
    y=1.05
)

plt.show()
```

#### how much did PTSD screening change from 2014 to 2023/4?

```python
# Put 2014 and 2023/4 percentages into separate columns
ptsd_change = age_ptsd_chart.pivot(
    index=["gender", "age_group"],
    columns="survey_year",
    values="percentage"
).reset_index()

ptsd_change
```

```python
# calculate the percentage-point change from 2014 → 2023/4.

# Calculate change from 2014 to 2023/4
ptsd_change["change"] = (
    ptsd_change["2023/4"] - ptsd_change["2014"]
)

ptsd_change
```


Algorithm note 

We use subtraction to calculate the change in PTSD screening percentages between 2014 and 2023/4.

- `ptsd_change["2023/4"]` selects the PTSD screening percentages from 2023/4.
- `ptsd_change["2014"]` selects the percentages from 2014.
- Subtracting 2014 from 2023/4 calculates the percentage-point change.
- `ptsd_change["change"]` creates a new column to store the calculated differences.

Observations:

- A positive change indicates an increase in PTSD screening percentage.
- A negative change indicates a decrease.
- A change of zero indicates no difference between survey years.
- The original percentages remain unchanged.
- The number of rows remains the same.

Important:

These changes are descriptive and do not automatically indicate statistically significant differences.


```python
# Create change chart
g = sns.catplot(
    data=ptsd_change,
    x="age_group",
    y="change",
    col="gender",
    kind="bar",
    height=5,
    aspect=1
)

# Add zero reference line
for ax in g.axes.flat:
    ax.axhline(
        0,
        linewidth=1
    )

# Labels
g.set_axis_labels(
    "Age Group",
    "Change in PTSD Screen Positive (percentage points)"
)

g.set_titles("{col_name}")

# Main title
g.fig.suptitle(
    "Change in PTSD Screening by Age Group: 2014 to 2023/4",
    fontsize=15,
    fontweight="bold",
    y=1.05
)

plt.show()
```


Algorithm note - Visualising changes in PTSD screening between survey years

We use `sns.catplot()` to create bar charts showing the percentage-point changes in PTSD screening between 2014 and 2023/4 across age groups and genders.

- `data=ptsd_change` uses the dataset containing calculated percentage-point changes.
- `x="age_group"` displays the age groups.
- `y="change"` displays the percentage-point difference between survey years.
- `col="gender"` creates a separate chart for each gender.
- `kind="bar"` produces bar charts.
- `height` and `aspect` control the chart dimensions.

Chart formatting:

- `for ax in g.axes.flat` loops through each gender chart.
- `ax.axhline(0)` adds a horizontal reference line at zero.
- `linewidth=1` controls the reference line thickness.
- `set_axis_labels()` adds descriptive axis labels.
- `set_titles("{col_name}")` displays the gender above each chart.
- `suptitle()` adds the overall chart title.
- `plt.show()` displays the completed visualisation.

The chart displays percentage-point changes, not relative percentage changes. These differences are descriptive and do not automatically indicate statistical significance.



Chart Interpretation - Changes in PTSD Screening by Age and Gender

The bar charts compare changes in PTSD screening percentages between 2014 and 2023/4 across seven age groups.

1. Overall adult population

- Ages 16-24 experienced the largest increase (+3.36 percentage points).
- Ages 25-34 had the second-largest increase (+2.63 percentage points).
- Increases generally became smaller among older age groups.
- Ages 75+ showed a slight decrease (-0.14 percentage points).

2. Men

- Men aged 16-24 experienced the largest increase (+6.75 percentage points).
- Men aged 25-34 also showed a substantial increase (+3.68 percentage points).
- Men aged 55-64 showed a decrease (-1.54 percentage points).
- Other male age groups experienced relatively small changes.

3. Women

- Women aged 55-64 experienced the largest increase (+2.99 percentage points).
- Women aged 35-44 increased by approximately 1.94 percentage points.
- Women aged 25-34 increased by approximately 1.63 percentage points.
- Women aged 16-24 remained almost unchanged (+0.03 percentage points).
- Women aged 75+ showed a slight decrease (-0.36 percentage points).

4. Gender differences

- The largest increase among men occurred at ages 16-24.
- The largest increase among women occurred at ages 55-64.
- Changes in PTSD screening percentages differed considerably by age and gender.
- The increase among young men was much larger than among young women.

Overall conclusion:

PTSD screening percentages generally increased between 2014 and 2023/4, but the changes were not uniform across age groups and genders.

Younger men experienced particularly large increases, while middle-aged women showed notable increases in some age groups.

These findings suggest that age and gender are important factors to consider when investigating changes in PTSD screening prevalence over time.

Important:

These are descriptive percentage-point changes, not evidence of causation or statistical significance. Further statistical analysis is required to determine whether the differences are statistically significant.
